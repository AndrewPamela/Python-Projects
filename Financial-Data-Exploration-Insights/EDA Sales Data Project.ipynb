```python
#Clean Missing data
# Region:
df ["Region"] = df ["Region"]. fillna("Unknown")

```

```python
# Customer_name:
df ["Customer_Name"] = df ["Customer_Name"]. fillna("Unknown")
```

```python
# Sales:
df["Sales"] = df["Sales"].fillna(0)
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
```

```python
# Check missing values again
df.isnull().sum()
```

```python
# Check for Duplicates
df.duplicated().sum()
```

```python
# Remove Duplicate
df=df.drop_duplicates()
```

```python
# Convert the Date
df["Date"] = pd.to_datetime(df["Date"])
```

```python
sns.barplot(data=df, x="Region", y="Sales", estimator="sum")
plt.title("Regional Sales Performance")
plt.xlabel("Region")
plt.ylabel("Overall Sales Performance")
plt.xticks(rotation=45)
plt.show()
```

```python
### Regional Sales Performance: This chart compares total sales across different regions. It helps identify the highest- and lowest-performing regions and provides insight into geographical sales distribution.
```

```python
sns.barplot(data=df, x="Product", y="Sales", estimator="sum")
plt.title("Sales Performance by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
### Sales Performance by Product: This chart compares total sales generated by each product. It helps identify the best-selling products and highlights products that may require further marketing or performance review.
```

```python
# Sales Trend Over Time
# First Aggregate sales by date:
```

```python
# Sales Performance Over Time:This analysis examines how sales change over time.

# Aggregate total sales by date

# The sales data is grouped by date, and the total sales for each day are calculated to identify trends and fluctuations over time.
```

```python
Daily_Sales = df.groupby("Date")["Sales"].sum()

    # Create the line chart

plt.figure(figsize=(12, 5))
plt.plot(Daily_Sales)
plt.title("Sales Performance Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

```python
# Question
# Is sales performance stable? No
# Are there periods of unusually high sales? yes
# Are there periods of unusually low sales? no
# can you identify any trend? yes
```

```python
### Sales Distribution
```

```python
### Correlation Analysis
```

```python
# Calculate the correlation
Correlation = df.corr(numeric_only=True)

# Visualize it
plt.figure(figsize=(8, 6))
sns.heatmap(Correlation,annot=True,cmap="coolwarm")

plt.title("Correlation Matrix")
plt.show()
```
