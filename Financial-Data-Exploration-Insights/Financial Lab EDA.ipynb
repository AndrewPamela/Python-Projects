```python
# Calculate total income
Total_Income = df[df["Transaction_Type"]=="Income"]["Amount"].sum()
```

```python
# Calculate total income
Total_Expense = df[df["Transaction_Type"]=="Expense"]["Amount"].sum()
```

```python
print(Total_Expense)
```

```python
#calculate the net cash flow
Net_cash_flow = Total_Income-Total_Expense
```

```python
print(Net_cash_flow)
# If it is positive, it means the Income is greter than the Expenses but if it is negative the Expenses is gretater than the income
```

```python
# Calculate the Income to Expenses Ratio 
Total_Income/Total_Expense
```

```python
# Create a financial Summary
```

```python
Financial_Summary= pd.DataFrame({"Metrics":["Total_Income","Total_Expense","Net_cash_flow",
                                           "Average_Transaction","Maximun_Transaction","Minimum_Transaction"],
                                 "Amount":[Total_Income,Total_Expense,Net_cash_flow,
                                           df["Amount"].mean(),
                                           df["Amount"].max(),
                                           df["Amount"].min()
                                          ]})
```
