```python
# Identified:
# Columns containing missing values: Region, Customer_Segment, Sales
# Number of missing values: Region (30), Customer_Segment (20), Sales (26)

## Initial Data Findings: The dataset contains missing values in the Region, Customer_Segment, and Sales columns.
```

```python
### Calculate Missing Value Percentage: The percentage of missing values was calculated for each column in the dataset.
```

```python
# Calculate percentage of the missing values
(df.isnull().sum() / len(df)) * 100
```

```python
# Clean Missing Data

# Region:
df["Region"] = df["Region"].fillna("Unknown")
```

```python
## Clean Missing Data: The missing values in the Region column were replaced with “Unknown”.
```

```python
# Customer Segment:
df["Customer_Segment"] = df["Customer_Segment"].fillna("Unknown")
```

```python
## Clean Customer Segment: The missing values in the Customer_Segment column were replaced with “Unknown”.
```

```python
# Sales:
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
```

```python
## Clean Sales Data: The missing values in the Sales column were replaced with the median sales value.
```

```python
# Check missing values again
df.isnull().sum()
```

```python
## Verify Missing Values: The dataset was checked again to confirm that the missing values had been successfully handled.
```

```python
# Check for Duplicates
df.duplicated().sum()
```

```python
## Check for Duplicates: The dataset was checked for duplicate records to ensure that each record appears only once.
```

```python
# Remove Duplicates
df = df.drop_duplicates()
```
