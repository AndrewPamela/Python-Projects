```python
# Sales Performance Analysis

sales_summary = df.groupby(["Region", "Customer_Segment"]).agg(
    Total_Sales=("Sales", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Average_Sales=("Sales", "mean"),
    Average_Discount=("Discount", "mean")
).reset_index()

sales_summary = sales_summary.sort_values(
    by="Total_Sales",
    ascending=False
)

print(sales_summary)
```

```python
# Visualize Sales by Region
plt.figure(figsize=(8, 5))
sns.barplot(
    data=sales_summary,
    x="Region",
    y="Total_Sales",
    estimator="sum")
plt.title("Sales Performance by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot: Quantity vs Sales
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="Quantity",
    y="Sales",
    hue="Customer_Segment",
    s=80)
plt.title("Relationship Between Quantity and Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```
